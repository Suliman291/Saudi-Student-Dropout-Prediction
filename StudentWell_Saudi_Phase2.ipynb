Run Step 1 to Step 12 in order.
The test set is kept untouched until the final evaluation.


## Step 1 — Install packages and load the CSV


```python
!pip -q install imbalanced-learn

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from google.colab import files

uploaded = files.upload()
filename = next(iter(uploaded))

df = pd.read_csv(filename)

# Clean column names
df.columns = df.columns.astype(str).str.strip()

print("Dataset shape:", df.shape)
display(df.head())
print("\nColumns:")
print(df.columns.tolist())
```

## Step 2 — Inspect and clean the dataset


```python
print("Original rows:", len(df))
print("Duplicate rows:", df.duplicated().sum())
print("Total missing values:", df.isnull().sum().sum())

# Remove exact duplicate rows
df = df.drop_duplicates().reset_index(drop=True)

print("\nRows after removing duplicates:", len(df))
print("\nData types:")
print(df.dtypes.value_counts())

display(df.head())
```

## Step 3 — Detect the target and inspect class distribution


```python
# Find the Dropout target column safely
target_matches = [
    c for c in df.columns
    if c.strip().lower() == "dropout"
]

if not target_matches:
    raise ValueError(
        "Could not find a column named 'Dropout'. "
        "Check the printed column names in Step 1."
    )

target_col = target_matches[0]

print("Target column:", target_col)
print("\nTarget values:")
print(df[target_col].value_counts(dropna=False))

print("\nTarget percentages:")
print(
    (df[target_col].value_counts(normalize=True) * 100)
    .round(2)
)
```

## Step 4 — Basic visualizations


```python
target_counts = df[target_col].value_counts()

plt.figure(figsize=(7, 4))
plt.bar(
    target_counts.index.astype(str),
    target_counts.values
)
plt.title("Student Dropout Distribution")
plt.xlabel("Dropout")
plt.ylabel("Number of Students")
plt.tight_layout()
plt.show()

# Show the most common values for selected important factors if available
important_names = [
    "GPA",
    "Academic year",
    "Employment status",
    "Family support",
    "Major pre-knowledge"
]

for keyword in important_names:
    matches = [
        c for c in df.columns
        if keyword.lower() in c.lower()
    ]

    if matches:
        c = matches[0]

        print(f"\n{c}:")
        display(
            df.groupby(c)[target_col]
            .value_counts(normalize=True)
            .mul(100)
            .round(1)
            .rename("Percent")
            .reset_index()
            .head(20)
        )
```

## Step 5 — Prepare X and y


```python
X = df.drop(columns=[target_col]).copy()
y = df[target_col].copy()

# Remove rows where target is missing, if any
valid = y.notna()
X = X.loc[valid].reset_index(drop=True)
y = y.loc[valid].reset_index(drop=True)

print("X shape:", X.shape)
print("y shape:", y.shape)
print("\nClass distribution:")
print(y.value_counts())
```

## Step 6 — Train/Test split


```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training rows:", len(X_train))
print("Testing rows:", len(X_test))

print("\nTraining distribution:")
print(y_train.value_counts())

print("\nTesting distribution:")
print(y_test.value_counts())
```

## Step 7 — Preprocessing + SMOTE-ENN + models


```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline as SkPipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer

from imblearn.combine import SMOTEENN
from imblearn.pipeline import Pipeline as ImbPipeline

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC

# Detect numeric and categorical columns
numeric_cols = X_train.select_dtypes(
    include=["number"]
).columns.tolist()

categorical_cols = [
    c for c in X_train.columns
    if c not in numeric_cols
]

print("Numeric columns:", len(numeric_cols))
print("Categorical columns:", len(categorical_cols))

numeric_transformer = SkPipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_transformer = SkPipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(
        handle_unknown="ignore",
        sparse_output=False
    ))
])

preprocessor = ColumnTransformer([
    ("num", numeric_transformer, numeric_cols),
    ("cat", categorical_transformer, categorical_cols)
])

def make_model_pipeline(model):
    return ImbPipeline([
        ("preprocess", preprocessor),
        ("balance", SMOTEENN(
            sampling_strategy=0.7,
            random_state=42
        )),
        ("model", model)
    ])

logistic_model = make_model_pipeline(
    LogisticRegression(
        max_iter=3000,
        random_state=42
    )
)

forest_model = make_model_pipeline(
    RandomForestClassifier(
        n_estimators=300,
        random_state=42,
        n_jobs=-1
    )
)

svm_model = make_model_pipeline(
    SVC(
        kernel="rbf",
        random_state=42
    )
)
```

## Step 8 — Initial 5-fold cross-validation


```python
from sklearn.model_selection import StratifiedKFold, cross_validate

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

models = {
    "Logistic Regression": logistic_model,
    "Random Forest": forest_model,
    "SVM": svm_model
}

rows = []

for name, model in models.items():
    scores = cross_validate(
        model,
        X_train,
        y_train,
        cv=cv,
        scoring=["accuracy", "f1_macro"],
        n_jobs=-1
    )

    rows.append({
        "Model": name,
        "CV Accuracy (%)": round(
            scores["test_accuracy"].mean() * 100, 2
        ),
        "CV Macro F1": round(
            scores["test_f1_macro"].mean(), 3
        )
    })

cv_results = pd.DataFrame(rows).sort_values(
    "CV Macro F1",
    ascending=False
)

display(cv_results)
```

## Step 9 — Hyperparameter tuning


```python
from sklearn.model_selection import GridSearchCV

logistic_grid = GridSearchCV(
    logistic_model,
    param_grid={
        "model__C": [0.1, 1, 10],
        "model__class_weight": [None, "balanced"]
    },
    cv=cv,
    scoring={
        "accuracy": "accuracy",
        "f1_macro": "f1_macro"
    },
    refit="f1_macro",
    n_jobs=-1
)

forest_grid = GridSearchCV(
    forest_model,
    param_grid={
        "model__n_estimators": [300, 500],
        "model__max_depth": [8, 15, None],
        "model__min_samples_leaf": [1, 2, 4],
        "model__max_features": ["sqrt", "log2"],
        "model__class_weight": [None, "balanced"]
    },
    cv=cv,
    scoring={
        "accuracy": "accuracy",
        "f1_macro": "f1_macro"
    },
    refit="f1_macro",
    n_jobs=-1
)

svm_grid = GridSearchCV(
    svm_model,
    param_grid={
        "model__C": [0.1, 1, 10],
        "model__gamma": ["scale", 0.01, 0.1],
        "model__class_weight": [None, "balanced"]
    },
    cv=cv,
    scoring={
        "accuracy": "accuracy",
        "f1_macro": "f1_macro"
    },
    refit="f1_macro",
    n_jobs=-1
)

logistic_grid.fit(X_train, y_train)
forest_grid.fit(X_train, y_train)
svm_grid.fit(X_train, y_train)

tuned_models = {
    "Logistic Regression": logistic_grid,
    "Random Forest": forest_grid,
    "SVM": svm_grid
}

tuned_rows = []

for name, grid in tuned_models.items():
    best_i = grid.best_index_

    tuned_rows.append({
        "Model": name,
        "Best CV Accuracy (%)": round(
            grid.cv_results_["mean_test_accuracy"][best_i] * 100,
            2
        ),
        "Best CV Macro F1": round(
            grid.cv_results_["mean_test_f1_macro"][best_i],
            3
        )
    })

tuned_results = pd.DataFrame(
    tuned_rows
).sort_values(
    "Best CV Macro F1",
    ascending=False
)

display(tuned_results)

print("\nBest Logistic settings:")
print(logistic_grid.best_params_)

print("\nBest Random Forest settings:")
print(forest_grid.best_params_)

print("\nBest SVM settings:")
print(svm_grid.best_params_)
```

## Step 10 — Select the best model using CV, then test once


```python
from sklearn.metrics import (
    accuracy_score,
    f1_score,
    classification_report,
    ConfusionMatrixDisplay
)

best_model_name = max(
    tuned_models,
    key=lambda name: tuned_models[name].best_score_
)

best_model = tuned_models[
    best_model_name
].best_estimator_

test_predictions = best_model.predict(X_test)

test_accuracy = accuracy_score(
    y_test,
    test_predictions
)

test_macro_f1 = f1_score(
    y_test,
    test_predictions,
    average="macro"
)

print("Selected Model:", best_model_name)
print(
    "Test Accuracy:",
    round(test_accuracy * 100, 2),
    "%"
)
print(
    "Test Macro F1:",
    round(test_macro_f1, 3)
)

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        test_predictions,
        zero_division=0
    )
)

ConfusionMatrixDisplay.from_predictions(
    y_test,
    test_predictions
)

plt.title(
    best_model_name +
    " - Final Confusion Matrix"
)
plt.tight_layout()
plt.show()
```

## Step 11 — Feature importance


```python
from sklearn.inspection import permutation_importance

importance = permutation_importance(
    best_model,
    X_test,
    y_test,
    scoring="f1_macro",
    n_repeats=5,
    random_state=42,
    n_jobs=-1
)

feature_importance = pd.DataFrame({
    "Feature": X.columns,
    "Importance": importance.importances_mean
}).sort_values(
    "Importance",
    ascending=False
)

print("Top 10 important original features:")
display(feature_importance.head(10))

top10 = (
    feature_importance
    .head(10)
    .sort_values("Importance")
)

plt.figure(figsize=(9, 5))
plt.barh(
    top10["Feature"],
    top10["Importance"]
)
plt.title(
    "Top 10 Features - " +
    best_model_name
)
plt.xlabel("Permutation Importance")
plt.tight_layout()
plt.show()
```

## Step 12 — Final summary


```python
final_results = pd.DataFrame({
    "Selected Model": [best_model_name],
    "Test Accuracy (%)": [
        round(test_accuracy * 100, 2)
    ],
    "Test Macro F1": [
        round(test_macro_f1, 3)
    ]
})

print("FINAL RESULTS")
display(final_results)

if test_accuracy >= 0.80:
    print("Accuracy requirement reached: >= 80%")
else:
    print("Accuracy is below 80%. Do not change the test set or add target-leaking features.")
```
